### basic methods -

```python
k = 1,2,3,3,4,5
r = set(k)

print(r)
print(type(r))
```

```python
k = {1,2,3,3,3,4,5}
k.add(25)

print(k)
```

```python
k = {1,2,3,3,3,4,5}
r = [25,50,75]

k.update(r)
print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.remove(25)

print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.remove(75)

print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.discard(25)

print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.discard(75)

print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.pop()

print(k)
```

```python
k = {1, 2, 3, 4, 5, 25}
k.clear()

print(k)
```

#### for SET operations -

[
\begin{array}{|l|l|l|l|}
\hline
\textbf{Operation} & \textbf{Description} & \textbf{Method Syntax} & \textbf{Operator Syntax} \\
\hline
\textbf{Union} &
\text{Combines all unique elements from both sets.} &
\texttt{set1.union(iterable)} &
\texttt{set1 \textbackslash | set2} \\
\hline
\textbf{Intersection} &
\text{Extracts only the elements common to both sets.} &
\texttt{set1.intersection(iterable)} &
\texttt{set1 \& set2} \\
\hline
\textbf{Difference} &
\text{Extracts elements in the first set that are not in the second.} &
\texttt{set1.difference(iterable)} &
\texttt{set1 - set2} \\
\hline
\end{array}
]

```python
x = {1, 2, 3, 4, 5, 6, 7}
y = {4, 5, 6, 7, 8, 9, 10}
```

```python
x.union(y)
```

```python
x|y
```

```python
x.intersection(y)
```

```python
x&y
```

```python
x.difference(y)
```

```python
x-y
```
