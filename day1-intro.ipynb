```python
x = y = z = 5

print(x, y, z)
print(id(x), id(y), id(z))
```

##### **Note:** Here, all ID's are same because x, y and z are pointing to same memory block.

```python
z = 10

print(x, y, z)
print(id(x), id(y), id(z))
```

##### **Note:** Here, x, y ID's are same but z is different.


```python
x, y = 5, 10

#before swapping
print(f"x={x}, y={y}")

x,y = y, x

#after swapping
print(f"x={x}, y={y}")
```
