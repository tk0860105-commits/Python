```python
#WAP to check odd or even number.
x = int(input("Enter a number: "))
if(x%2==0):
    print(f"The {x} is even.")
else:
    print(f"The {x} is odd.")
```

```python
#WAP to check if a number is positive, negative or zero.
x = int(input("Enter a number: ")) 

if(x>0):
    print(f"The {x} is positive.")
elif(x<0):
    print(f"The {x} is negative.")
else:
    print(f"The {x} is zero.")
```

```python
#Wap to check vowel or consonant.
x = input("Enter a character: ")
if(x=='a' or x=='e' or x=='i' or x=='o' or x=='u'):
    print(f"The {x} is vowel.")
else:
    print(f"The {x} is consonant.")
```

```python
#Wap to Check leap year or not.
x = int(input("Enter a year: "))
if(x%4==0 and x%100!=0 or x%400==0):
    print(f"The {x} is a leap year.")
else:
    print(f"The {x} is not a leap year.")
```
