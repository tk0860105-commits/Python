```python
#Wap to Check leap year or not.
x = int(input("Enter a year: "))
if(x%4==0 and x%100!=0 or x%400==0):
    print(f"The {x} is a leap year.")
else:
    print(f"The {x} is not a leap year.")
```

```python
#Wap to impliment match-case to find square and cube.

print("menu -")
print(" 1. square")
print(" 2. cube")

x = int(input("Enter any value"))
ch = int(input("Enter your choice"))

match(ch):
    case 1:
        print(f"Square of {x} = {x*x}")
    case 2:
        print(f"Cube of {x} = {x*x*x}")
    case _:
        print("Wrong choice")
```

```python
#Wap to impliment match-case to .

print("menu -")
print(" 1. Addition")
print(" 2. Subtraction")
print(" 3. Division")
print(" 4. Multiplication")

x = int(input("Enter First value"))
y = int(input("Enter Second value"))
ch = int(input("Enter your choice"))

match(ch):
    case 1:
        print(f"Addition of {x} and {y} = {x+y}")
    case 2:
        print(f"Subtraction of {x} and {y} = {x-y}")
    case 3:
        print(f"Division of {x} and {y} = {x/y}")
    case 4:
        print(f"Multiplication  of {x} and {y} = {x*y}")
    case _:
        print("Wrong choice")
```
