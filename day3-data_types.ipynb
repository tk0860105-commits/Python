```python
# 1-boolean data type
t = True
f = False
print(type(t))
print(type(f))  
```

```python
# 2-number data type
i = 5
f = 3.14
complex_number = 2 + 3j
print(type(i))
print(type(f))
print(type(complex_number))
```

```python
# 3-sequence data type
n1, n2 = 'Tarun', "Kumar"
my_list = [1, 2, 3, 4]
my_tuple = (1, 2, 3, 4)
print(type(n1),type(n2))
print(type(my_list))
print(type(my_tuple))   
```

```python
# 4-set data type
my_set = {1, 2, 3, 3, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
# 5-dictionary data type
my_dict = {'name': 'Tarun', 'age': 25, 'city': 'Kanpur'}
print(my_dict)
print(type(my_dict))
```
