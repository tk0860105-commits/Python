### list()

```python
k = 1,2,3,4,5
r = list(k)
print(k)
print(r)
```

### index(), count(), copy(), len()

```python
k = [1,2,3,4,5]
print(k.index(5))
print(k.count(2))

r = k.copy()
print(r)

print(len(k))
print(k)

```

### insert()

```python
k = [10,20,30,40]
k.insert(2,50)
print(k)
```

```python
k.remove(10)
print(k)
```

```python
k = [10,20,30,40,50]
k.append(50)
print(k)
```

```python
k = [10,20,30]
r = [30,40,50]
k.extend(r)
print(k)
```

```python
k = [10,20,30,40]
k.pop()
print(k)

k.pop(1)
print(k)
```

```python
k = [10,20,30,40,50,60]
k.clear()
print(k)
```

```python
k = [10,20,30,40,50,60]
del k
print(k)
```
