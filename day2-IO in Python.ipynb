#### How to accept input and print output in Python?


```python
# If you want to accept any input and print it.
x = input("Enter any value: ")
print(x)
```

```python
# If you want to input any users name and age and print it.
name = input("Enter your name: ")
age = input("Enter your age: ")
print("Name: " + name)
print("Age: " + age)
```

```python
# By f string method.
name = input("Enter your name: ")
age = input("Enter your age: ")
print(f"Your name is {name} and you are {age} years old.")
```

```python
# By format method.
name = input("Enter your name: ") 
age = input("Enter your age: ")
print("Your name is {0} and you are {1} years old.".format(name, age))
```
