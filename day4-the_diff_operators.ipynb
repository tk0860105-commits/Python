#### We have following type of operators-
1. Arithmetic Operators
2. Comparison Operators
3. Assignment Operators
4. Logical Operators 
5. Bitwise Operators
6. Identity Operators and Membership Opeartors

```python
#1) Arithmetic Operators -
x,y = 5,10
print("Addition:", x+y) 
print("Subtraction:", x-y) 
print("Multiplication:", x*y) 
print("Division:", x/y) 
print("Modulus:", x%y) 
print("Exponentiation:", x**y) 
print("Floor Division:", x//y) 
```

```python
#2) Comparison Operators -
x,y = 5,10
print("Equal:", x == y)
print("Not Equal:", x != y)
print("Greater Than:", x > y)
print("Less Than:", x < y)
print("Greater Than or Equal:", x >= y)
print("Less Than or Equal:", x <= y)
```

```python
#3) Assignment Operators -
x = 5
print("x =", x)
x += 3  # x = x + 3
print("x += 3:", x)
x -= 2  # x = x - 2
print("x -= 2:", x)
x *= 4  # x = x * 4
print("x *= 4:", x)
x /= 2  # x = x / 2
print("x /= 2:", x)
```

```python
#4) Logical Operators -
x = True
y = False
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
```

```python
#5) Bitwise Operators -
x = 5
y = 3
print("x & y (AND):", x & y)
print("x | y (OR):", x | y)
print("x ^ y (XOR):", x ^ y)
print("~x (NOT):", ~x)
print("x << 1 (Left Shift):", x << 1)
print("x >> 1 (Right Shift):", x >> 1)
```

```python
#6) Identity Operators and Membership Operators-
x = 5
y = 10
print("x is y:", x is y)
print("x is not y:", x is not y)

k = [1, 2, 3]
print("5 in k:", 5 in k)
print("2 in k:", 2 in k)
print("4 not in k:", 4 not in k)
```

#### The Operator Precedence in Python -

| Precedence | Operator | Description |
|------------|----------|-------------|
| 1 | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x`, `~x` | Unary Plus, Unary Minus, Bitwise NOT |
| 4 | `*`, `/`, `//`, `%` | Multiplication, Division, Floor Division, Modulus |
| 5 | `+`, `-` | Addition, Subtraction |
| 6 | `<<`, `>>` | Bitwise Shift |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `\|` | Bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=` | Comparison |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 | `or` | Logical OR |
