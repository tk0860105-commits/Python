### 1) If you want to create any file –

```python
f = open("myfile.txt", "x")
f.write("This is my first file!")
print(f"File '{f}' has been created.")
f.close()
```

### 2) If you want to read any data from any file –

```python
f = open("myfile.txt")
print(f.read())
f.close()
```

### 3) If you want to write any data to any file –

```python
f = open("myfile.txt", "w+")
f.write("This is my another text in this file!")
f.close()
```

### 4) If you want to remove any file –

```python
import os
os.remove("myfile.txt")
```

```python
try:
    # Attempt to open and read the file
    with open("example.txt", "r") as file:
        content = file.read()
        print("File content:\n", content)

except FileNotFoundError:
    print("Error: The file does not exist!")

except PermissionError:
    print("Error: You don’t have permission to access this file.")

except Exception as e:
    print("An unexpected error occurred:", e)

finally:
    print("File operation complete.")
```

```python
try:
    # Attempt to write to the file
    with open("output.txt", "w") as file:
        file.write("Hello, Python file handling with exceptions!")

except PermissionError:
    print("Error: Cannot write to the file due to permission issues.")

except Exception as e:
    print("Unexpected error:", e)

finally:
    print("Write operation finished.")
```
