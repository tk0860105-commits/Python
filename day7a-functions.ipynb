```python
#WAP to create a function to find square and cube of a number
def calc(a):
    square = a**2
    cube = a**3
    return square,cube

x = input("Enter any value:")
square, cube = calc(int(x))
print("square: ", square)
print("cube: ", cube)
```

```python
#Wap to create a function to find Area and Circumference of a circle.
def calc(r):
    area = 3.14*r*r
    circumference = 2*3.14*r
    return area,circumference

x = input("Enter radius = ")
area, circumference = calc(int(x))
print("Area of circle: ",area)
print("circumference of circle: ",circumference)
    
```

```python
#Wap to create a function to find Simple Interest and compound Interest.
def calc(p,r,t):
    si = (p*r*t)/100
    A = p*pow(1+r/100,t)
    ci = A-p
    return A, si, ci

print("Enter the principal, rate and time:")
p = int(input())
r = float(input())
t = int(input())
A, si, ci = calc(p,r,t)
print("A:", A)
print("Simple Interest = ",si)
print("Compound Interest = ",ci)
```
