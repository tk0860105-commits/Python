```python
#Wap to find Armstong Number.
n = int(input("Enter any number : "))
count = 0
sum = 0
while n > 0:
    n = n//10
    count = count+1
print("Length = ",count)
for i in range (count):
    rem = n%10
    sum = sum + rem ** count
    rem // 10

if sum == n:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
#Wap to find Armstong Number.
n = int(input("Enter the number : "))
temp = n
sum = 0
fact = 1

while n > 0:
    rem = n % 10
    i=1
    while i <= rem:
        fact = fact*i
        i += 1

    sum = sum + fact
    n = n//10
    fact=1

if sum == temp:
    print(temp,"is strong Number.")
else:
    print(temp,"is Not strong Number.")
```

```python
#Wap to find palindrome.
n = int(input("Enter any number : "))
temp = n
rev = 0
while n > 0:
    r = n%10
    rev = rev*10 + r
    n = n//10

if rev == temp:
    print(temp,"is a Palindrome number.")
else:
    print(temp,"is a Not Palindrome number.")
```
